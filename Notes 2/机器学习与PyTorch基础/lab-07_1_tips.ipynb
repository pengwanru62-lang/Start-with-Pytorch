# 实验 7-1：技巧 (Tips)

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
# 保证结果可复现
torch.manual_seed(1)
```

## 训练集与测试集

```python
x_train = torch.FloatTensor([[1, 2, 1],
                             [1, 3, 2],
                             [1, 3, 4],
                             [1, 5, 5],
                             [1, 7, 5],
                             [1, 2, 5],
                             [1, 6, 6],
                             [1, 7, 7]
                            ])
y_train = torch.LongTensor([2, 2, 2, 1, 1, 1, 0, 0])
```

```python
x_test = torch.FloatTensor([[2, 1, 1], [3, 1, 2], [3, 3, 4]])
y_test = torch.LongTensor([2, 2, 2])
```

## 模型

```python
class SoftmaxClassifierModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(3, 3)
    def forward(self, x):
        return self.linear(x)
```

```python
model = SoftmaxClassifierModel()
```

```python
# 优化器设置
optimizer = optim.SGD(model.parameters(), lr=0.1)
```

```python
def train(model, optimizer, x_train, y_train):
    nb_epochs = 20
    for epoch in range(nb_epochs):

        # 计算预测值 H(x)
        prediction = model(x_train)
        # 计算代价 (Cost)
        cost = F.cross_entropy(prediction, y_train)

        # 反向传播与参数更新
        optimizer.zero_grad()
        cost.backward()
        optimizer.step()

        print('Epoch {:4d}/{} Cost: {:.6f}'.format(
            epoch, nb_epochs, cost.item()
        ))
```

```python
def test(model, optimizer, x_test, y_test):
    prediction = model(x_test)
    predicted_classes = prediction.max(1)[1]
    correct_count = (predicted_classes == y_test).sum().item()
    cost = F.cross_entropy(prediction, y_test)

    print('Accuracy: {}% Cost: {:.6f}'.format(
         correct_count / len(y_test) * 100, cost.item()
    ))
```

```python
train(model, optimizer, x_train, y_train)
```

```python
test(model, optimizer, x_test, y_test)
```

## 学习率 (Learning Rate)

梯度下降 (Gradient Descent) 中的 $\alpha$ 值

在 `optimizer = optim.SGD(model.parameters(), lr=0.1)` 中，`lr=0.1`。

如果学习率太大，可能会导致发散 (diverge)，从而使代价 (cost) 逐渐增加 (过冲/overshooting)。

```python
model = SoftmaxClassifierModel()
```

```python
optimizer = optim.SGD(model.parameters(), lr=1e5)
```

```python
train(model, optimizer, x_train, y_train)
```

如果学习率太小，代价 (cost) 几乎不会减少。

```python
model = SoftmaxClassifierModel()
```

```python
optimizer = optim.SGD(model.parameters(), lr=1e-10)
```

```python
train(model, optimizer, x_train, y_train)
```

以一个适当的数值开始，如果发散就减小它，如果代价不减少就增大它。

```python
model = SoftmaxClassifierModel()
```

```python
optimizer = optim.SGD(model.parameters(), lr=1e-1)
```

```python
train(model, optimizer, x_train, y_train)
```

## 数据预处理 (Data Preprocessing)

让我们对数据进行零中心化 (zero-center) 和归一化 (normalize)。

```python
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 90],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

$$ x'_j = \frac{x_j - \mu_j}{\sigma_j} $$

这里 $\sigma$ 是标准差，$\mu$ 是平均值。

```python
mu = x_train.mean(dim=0)
```

```python
sigma = x_train.std(dim=0)
```

```python
norm_x_train = (x_train - mu) / sigma
```

```python
print(norm_x_train)
```

让我们使用归一化和零中心化后的 X 进行训练，看看性能如何。

```python
class MultivariateLinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(3, 1)

    def forward(self, x):
        return self.linear(x)
```

```python
model = MultivariateLinearRegressionModel()
```

```python
optimizer = optim.SGD(model.parameters(), lr=1e-1)
```

```python
def train(model, optimizer, x_train, y_train):
    nb_epochs = 20
    for epoch in range(nb_epochs):

        # 计算预测值 H(x)
        prediction = model(x_train)

        # 计算代价 (Cost)
        cost = F.mse_loss(prediction, y_train)

        # 反向传播与参数更新
        optimizer.zero_grad()
        cost.backward()
        optimizer.step()

        print('Epoch {:4d}/{} Cost: {:.6f}'.format(
            epoch, nb_epochs, cost.item()
        ))
```

```python
train(model, optimizer, norm_x_train, y_train)
```

## 过拟合 (Overfitting)

如果模型在训练数据上学习得太好，可能导致在测试数据上表现不佳。

防止这种情况的方法主要有三种：

1. 更多的训练数据
2. 更少的特征 (feature)
3. **正则化 (Regularization)**

正则化：不要让权重 (weights) 的数值太大

```python
def train_with_regularization(model, optimizer, x_train, y_train):
    nb_epochs = 20
    for epoch in range(nb_epochs):

        # 计算预测值 H(x)
        prediction = model(x_train)

        # 计算代价 (Cost)
        cost = F.mse_loss(prediction, y_train)
        
        # 计算 L2 正则化项
        l2_reg = 0
        for param in model.parameters():
            l2_reg += torch.norm(param)
            
        cost += l2_reg

        # 反向传播与参数更新
        optimizer.zero_grad()
        cost.backward()
        optimizer.step()

        print('Epoch {:4d}/{} Cost: {:.6f}'.format(
            epoch+1, nb_epochs, cost.item()
        ))
```

```python
model = MultivariateLinearRegressionModel()
```

```python
optimizer = optim.SGD(model.parameters(), lr=1e-1)
```

```python
train_with_regularization(model, optimizer, norm_x_train, y_train)
```
