## 从 .csv 文件加载数据 (Loading Data from `.csv` file)

```python
import numpy as np
```

```python
xy = np.loadtxt('data-01-test-score.csv', delimiter=',', dtype=np.float32)
```

delimiter=','说明每一列数据是用逗号分隔的，那么每一行会被拆成  个数字。

```python
print(xy)
```

```python
x_data = xy[:, 0:-1]
y_data = xy[:, [-1]]
print(x_data)
print(y_data)
```

```python
print(x_data.shape) # x_data shape
print(len(x_data))  # x_data 长度
print(x_data[:5])   # 前五个
```

```python
print(y_data.shape) # y_data shape
print(len(y_data))  # y_data 长度
print(y_data[:5])   # 前五个
```

## 导入库

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
# 为了可复现性
torch.manual_seed(1)
```

## 低级实现 (Low-level Implementation)

```python
# 数据
x_train = torch.FloatTensor(x_data)
y_train = torch.FloatTensor(y_data)
# 模型初始化
W = torch.zeros((3, 1), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# 优化器设置
optimizer = optim.SGD([W, b], lr=1e-5)

nb_epochs = 20
for epoch in range(nb_epochs + 1):
    
    # 计算 H(x)
    hypothesis = x_train.matmul(W) + b # or .mm or @

    # 计算 cost
    cost = torch.mean((hypothesis - y_train) ** 2)

    # 通过 cost 改善 H(x)
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 每 100 次打印日志
    print('Epoch {:4d}/{} Cost: {:.6f}'.format(
        epoch, nb_epochs, cost.item()
    ))
```

## 使用 `nn.Module` 的高级实现 (High-level Implementation with `nn.Module`)

```python
class MultivariateLinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(3, 1)

    def forward(self, x):
        return self.linear(x)
```

```python
# 数据
x_train = torch.FloatTensor(x_data)
y_train = torch.FloatTensor(y_data)
# 模型初始化
model = MultivariateLinearRegressionModel()
# 优化器设置
optimizer = optim.SGD(model.parameters(), lr=1e-5)

nb_epochs = 20
for epoch in range(nb_epochs+1):
    
    # 计算 H(x)
    prediction = model(x_train)
    
    # 计算 cost
    cost = F.mse_loss(prediction, y_train)
    
    # 通过 cost 改善 H(x)
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()
    
    # 每 20 次打印日志
    print('Epoch {:4d}/{} Cost: {:.6f}'.format(
        epoch, nb_epochs, cost.item()
    ))
```

## 数据集与数据加载器 (Dataset and DataLoader)

如果数据太大，我们不能一次性加载所有的 `x_data` 和 `y_data`，只能按需加载必要的批次 (batch)。

# Dataset

**是什么：**

一个“数据访问接口”。

**作用：**

定义：
1. 数据一共有多少个样本
2. 如何获取第 i 个样本

**必须实现的两个方法：**

- `__len__()`      → 返回样本总数  
- `__getitem__(i)` → 返回第 i 个样本 (x, y)

**本质：**

一个索引到样本的映射：

    i → (x_i, y_i)


# DataLoader

**是什么：**

一个“数据批量迭代器”。

**作用：**

把 Dataset 中的样本：
- 按 batch_size 分批
- 可选打乱 (shuffle)
- 迭代输出给模型

**使用方式：**

    for x, y in dataloader:

**本质：**

把单个样本：

    (x_i, y_i)

组织成批数据：

    [(x_1,y_1), ..., (x_b,y_b)]


# 二者关系

Dataset：定义“数据长什么样”  
DataLoader：定义“数据怎么被送进模型”
